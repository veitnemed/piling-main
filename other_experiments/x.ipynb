```python
import numpy as np
from src.configs.config import init_rng
from src.functions import get_tau_array, calculate_complex_stiring_vector

f0 = float(input("Введите частоту, Гц: "))

params = init_rng()
params["f0"] = f0
tau = get_tau_array(params)

exponents = np.exp(-2j * np.pi * f0 * tau)
steering = calculate_complex_stiring_vector(
    params["target_phi"], tau, f0, params["M"]
)

for i, (delay, value) in enumerate(zip(tau, exponents)):
    print(
        f"Датчик {i}: τ = {delay * 1000:+.4f} мс; "
        f"exp(-j2πfτ) = {value.real:+.6f}{value.imag:+.6f}j;"
    )

print("\nСтиринг-вектор (строка — датчик; столбцы — 3 канала):")
print(steering.reshape(params["M"], 3))

ratio = params["c"] / (2 * params["R"] * f0)
phase_spread = 360 * f0 * np.ptp(tau)
limit = params["c"] / (20 * params["R"])

print(f"\nλ/(2R) = {ratio:.3f}")
print(f"Максимальная разность фаз датчиков = {phase_spread:.2f}°")
print(f"Граница условия λ/(2R) > 10: f < {limit:.2f} Гц")
```
