```python
%load_ext autoreload
%autoreload 2

import numpy as np
import time
from src.experiments import (stat_vs_params, calculate_mean, calculate_variance)
from src.functions import (calculate_barlet_values,  get_sin_signal, get_impulse_signal, calculate_stiring_matrix, calculate_complex_stiring_matrix, calculate_full_barlet_values,)
from src.configs.config import init_rng, restart_rng
import src.configs.config as cfg
import src.configs.comon as comon
from src.plotting import plot_variance_vs_param, plot_mean_vs_param
import os

origin_config = init_rng()
sin_signal = get_sin_signal
path = "attachment/images/report"
os.makedirs(path,exist_ok=True)
```

```python
config_parametrs1 = restart_rng(origin_config)
config_parametrs2 = restart_rng(origin_config)
start_time = time.perf_counter()

var_barlett_vs_phis = stat_vs_params(config_params=config_parametrs1,
                                   energy_func=calculate_barlet_values,
                                   signal_func = sin_signal,
                                   parametr_name = "target_phi",
                                   parametr_values = cfg.PHI_VALUES,
                                   stiring_func= calculate_stiring_matrix,
                                   is_CRN = False,
                                   stat_func = calculate_variance
                                   )

var_barlett_vs_phis_complex = stat_vs_params(config_params = config_parametrs2,
                                   energy_func = calculate_full_barlet_values,
                                   signal_func = sin_signal,
                                   parametr_name = "target_phi",
                                   parametr_values = cfg.PHI_VALUES,
                                   stiring_func= calculate_complex_stiring_matrix,
                                   is_CRN = False,
                                   stat_func = calculate_variance
                                   )

mean_barlett_vs_phis = stat_vs_params(config_params=config_parametrs1,
                                   energy_func=calculate_barlet_values,
                                   signal_func = sin_signal,
                                   parametr_name = "target_phi",
                                   parametr_values = cfg.PHI_VALUES,
                                   stiring_func= calculate_stiring_matrix,
                                   is_CRN = False,
                                   stat_func = np.mean
                                   )

mean_barlett_vs_phis_complex = stat_vs_params(config_params = config_parametrs2,
                                   energy_func = calculate_full_barlet_values,
                                   signal_func = sin_signal,
                                   parametr_name = "target_phi",
                                   parametr_values = cfg.PHI_VALUES,
                                   stiring_func= calculate_complex_stiring_matrix,
                                   is_CRN = False,
                                   stat_func = np.mean
                                   )

end_time = time.perf_counter()
delta_time = end_time - start_time
```

```python
plot_variance_vs_param(dict_values = {"Барлетт (без задержек)": var_barlett_vs_phis,
                                      "Барлетт (точная модель)": var_barlett_vs_phis_complex},
                       config_params = origin_config.copy(),
                       parametr_values = cfg.PHI_VALUES,
                       parametr_name = "phis",
                       delta_time = delta_time,
                       ylim = (0.00*comon.CONST_RADIAN, 0.1*comon.CONST_RADIAN),
                       xlim = (0,360),
                       type_signal = "garmonic"
                       )

plot_mean_vs_param(dict_values = {"Барлетт (без задержек)": mean_barlett_vs_phis,
                                      "Барлетт (точная модель)": mean_barlett_vs_phis_complex},
                       config_params = origin_config.copy(),
                       parametr_values = cfg.PHI_VALUES,
                       parametr_name = "phis",
                       delta_time = delta_time,
                       ylim = (-0.075*comon.CONST_RADIAN, 0.075*comon.CONST_RADIAN),
                       xlim = (0,360),
                       type_signal = "garmonic"
                       )
```
