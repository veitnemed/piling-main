```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
import time
from py_files.experiment_stats import (var_vs_params)
from py_files.functions import (get_barlet_values, get_cap_values, get_sin_signal, get_impulse_signal)
from py_files.configs.config import init_rng, restart_rng
import py_files.configs.config as cfg
import py_files.configs.comon as comon
from py_files.plot_scrip import plot_variance_vs_param
import os
origin_config = init_rng()
signal = get_sin_signal
path = "images/report"
os.makedirs(path,exist_ok=True)

f_vector = np.arange(1,100,1)
```

```python
def f_cos(
       R: float = 0.4,
       c: float = comon.SPEED_ON_SOUND,
       f_vector: np.ndarray = f_vector):

    return np.cos((2*np.pi*R*(1/c))*f_vector)


F1 = f_cos()

fig, ax = plt.subplots(1,1, figsize = (14,6))
ax.plot(f_vector,F1)
ax.set_xlabel(r"Центральная частота, Гц", labelpad=10, fontsize=20)
ax.set_ylabel(r"$\cos(2\pi R f/c)$", fontsize=18)

fig.suptitle(
    r"$F_1(f) = \cos(2\pi R f/c)$",
    y=0.925, va="center",
    fontfamily="Times New Roman", fontsize=22
)

ax.grid(True)


def f_sin(
       R: float = 0.4,
       c: float = comon.SPEED_ON_SOUND,
       f_vector: np.ndarray = f_vector):

    return np.sin((2*np.pi*R*(1/c))*f_vector)


F2 = f_sin()

fig, ax = plt.subplots(1,1, figsize = (14,6))
ax.plot(f_vector,F2)
ax.set_xlabel(r"Центральная частота, Гц", labelpad=10, fontsize=20)
ax.set_ylabel(r"$\sin(2\pi R f/c)$", fontsize=18)

fig.suptitle(
    r"$F_2(f) = \sin(2\pi R f/c)$",
    y=0.925, va="center",
    fontfamily="Times New Roman", fontsize=22
)

ax.grid(True)

```
