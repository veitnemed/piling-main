```python
sin_signal = get_sin_signal(config_parametrs)
```

```python
tau_array = get_tau_array(config_parametrs)
params_vector_copy = config_parametrs.copy()
```

```python
from logging import config


channel_matrix = get_channel_matrix(config_parametrs, tau_array, get_sin_signal)
W = get_covariation_matrix(channel_matrix, comon.N)
dir_matrix = calculate_stiring_matrix(config_parametrs)
dir_complex_matrix = calculate_complex_stiring_matrix(config_parametrs)

```

```python
E_bar_true = calculate_full_barlet_values(W,dir_complex_matrix)
E_bar = calculate_barlet_values(W, dir_matrix)
E_cap = calculate_capon_values(W, dir_matrix)
E_cap_true = calculate_full_capon_values(W,dir_complex_matrix)
```

```python
title_main = "Сравнение методов (Барлетт)"

snr_ratio = comon.SNR
energy_signal = calculate_energy(sin_signal,cfg.TIME_VECTOR)
sigma_noise = calculate_noise_level(energy_signal, snr_ratio, comon.DURATION_ALL)
phi_0 = comon.TARGET_PHI
phi_vector = cfg.PHI_VECTOR
A = comon.A

est_phi_true = get_est_phi(phi_vector, E_bar_true)
est_phi = get_est_phi(phi_vector, E_bar)

E_max_true = max(E_bar_true)
E_max = max(E_bar)

snr_text = (
    "∞ (без шума)" if np.isinf(snr_ratio)
    else f"{snr_ratio:.2f}"
)

fig, ax = plt.subplots(figsize = (14,10))
fig.subplots_adjust(top=0.85, bottom=0.28)

# Функция Бартлетта и отметки углов
ax.plot(
    phi_vector, E_bar_true,
    label=r"C учетом задержек"
)
ax.plot(
    phi_vector, E_bar,
    linestyle = "--",
    color = "red",
    label = "Без учёта задержек"
)



ax.scatter(est_phi_true, E_max_true, color="red", zorder=3)
ax.axhline(E_max_true, color="red", linestyle="--", alpha=0.4)

# Названия осей и заголовок
fig.suptitle(
    title_main,
    y=0.925, va="center",
    fontfamily="Times New Roman", fontsize=22
)

ax.set_xlabel(r"Проверяемый угол $\varphi$, рад.", labelpad=10, fontsize=20)
ax.set_ylabel(r"$E_{\mathrm{BAR}}(\varphi)$, Па$^2$", fontsize=18)

ax.set_xticks(
    [0, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi],
    ["0", r"$\pi/2$", r"$\pi$", r"$3\pi/2$", r"$2\pi$"]
)

ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(0, 1.15 * max(E_max_true, E_max))
ax.legend(loc="upper right", fontsize=12)
ax.grid(True)


plt.savefig(fname = f"attachment/images/notebook/{title_main}", bbox_inches = 'tight', dpi = 300)
plt.show()
rows = [
    [r"Метод: ", "Барлет", "----"],
    [r"Радиус АВП: ", f"{comon.RADIUS}", "м"],
    [r"Сигнал: ", r"Синус", "----"],
    [r"Частота: ", f"{comon.F_GARMONIC}", "Гц"],
    [r"SNR: ", f"{comon.SNR}", "----"],
    [r"Истиный угол $\varphi_{0}$ ", f"{(comon.TARGET_PHI*180)/np.pi:.2f}", "град"],
    [r"Оценка без задержек ", f"{(est_phi*180)/np.pi:.2f}", "град"],
    [r"С учетом задержек", f"{(est_phi_true*180)/np.pi:.2f}", "град"]]
create_and_save_table(rows, ["Параметр", "Значение", "Еденица"], f"attachment/images/notebook/table_{title_main}")
```
