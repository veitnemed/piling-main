```python
%load_ext autoreload
%autoreload 2

import numpy as np
import time
from py_files.experiment_stats import (var_vs_params)
from py_files.functions import (get_barlet_values, get_cap_values)
from py_files.config import init_rng, update_config
import py_files.config as cfg
from py_files.plot_scrip import plot_variance_vs_param

config_parametrs = init_rng()


```

```python
config_parametrs = update_config()

start_time = time.perf_counter()
    
        
var_barlett_vs_phis = var_vs_params(config_params=config_parametrs.copy(),
                                   energy_func=get_barlet_values,
                                   parametr_name = "target_phi",
                                   parametr_values = cfg.PHI_VALUES,
                                   )
var_cap_vs_phis = var_vs_params(config_params=config_parametrs.copy(),
                                   energy_func=get_cap_values,
                                   parametr_name = "target_phi",
                                   parametr_values = cfg.PHI_VALUES,
                                   )

end_time = time.perf_counter()
delta_time = end_time - start_time
```

```python
plot_variance_vs_param(dict_values = {"Барлетт": var_barlett_vs_phis,
                                      "Кейпон": var_cap_vs_phis},
                       config_params = config_parametrs.copy(),
                       parametr_values = cfg.PHI_VALUES,
                       parametr_name = "phis",
                       delta_time = delta_time,
                       ylim = (0, 1e-3)
                       )
```
